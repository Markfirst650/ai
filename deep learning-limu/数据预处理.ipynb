## 处理缺失数据

### 插值

```python
inputs,outputs = data_r.iloc[:,0:2],data_r.iloc[:,0:2]
inputs
```

**没有数值时，显示的是NaN，固定格式，不能改动**

***name.fillna填充NaN，name.mean()取平均值 带上numeric_only = True只对数字进行处理，否则会报错***

```python
inputs = inputs.fillna(inputs.mean(numeric_only = True))#fillna只能填充NaN
inputs
```

pd.get_dummies()：是 Pandas 中用于独热编码的函数，会将类别型特征（如Alley列的 “Pave”“NaN”）转换为多个二进制列（0/1）

dummy_na=True：表示将缺失值（NaN）也视为一个独立的类别，单独生成一列标识。

```python
inputs1 = pd.get_dummies(inputs,dummy_na = True)
inputs1
```
