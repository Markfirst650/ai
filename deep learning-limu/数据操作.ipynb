## tensor张量的链接  

**使用torch.cat**  
>dim=0是行连接 就是列数不变，叠加行   
dim=2是列连接  就是行数不变，叠加列

```python
x = torch.arange(12,dtype=torch.float32).reshape((3,4))#生成一个12个元素的张量，浮点类型，重组成3，4的矩阵格式
y = torch.tensor([[2.0,1,2,3],[1,2,3,4],[3,4,5,6]])
torch.cat((x,y),dim=0),torch.cat((x,y),dim=1)#要放到同一行且用逗号隔开才能显示两个结果
```
